```python
class Empty(Exception):
    pass
```

```python
class PriorityQueueQueue:
    def __init__(self):
        self.pq = UnsortedPriorityQueue()
        self.cont = 0

    def __len__(self):
        return len(self.pq)

    def is_empty(self):
        return len(self.pq) == 0
        
    def enqueue(self, value):
        self.cont +=1 
        self.pq.add(self.cont, value)

    def dequeue(self):
        if self.pq.is_empty():
            raise Empty('Queue is empty')
        key, value = self.pq.remove_min()
        return value
```

```python
class PriorityQueueStack:
    def __init__(self):
        self.pq = UnsortedPriorityQueue()
        self.cont = 0

    def is_empty(self):
        return len(self.pq) == 0

    def __len__(self):
        return len(self.pq)

    def push(self, value):
        self.cont -= 1
        self.pq.add(self.cont, value)

    def pop(self):
        if self.pq.is_empty():
            raise Empty('El stack está vacío')
        key, value = self.pq.remove_min()
        return value
    
```

```python
class StackPQ():
    def __init__(self):
        self.data = UnsortedPriorityQueue()
        self.key = 0

    def push(self, e):
        self.data.add(self.key, e)
        self.key -= 1

    def top(self):
        self.data.min()[1]

    def pop(self):
        if len(self.data) == 0:
            raise Empty('El stack está vacío')
        return self.data.remove_min()[1]
```
