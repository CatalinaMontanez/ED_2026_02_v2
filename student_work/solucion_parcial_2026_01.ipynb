```python
def Niidea(S):
    n = len(S)
    total = 0

    for j in range(0,n,2):
        for i in range(0,2):
            total = total + 2*S[j]*S[i]
    return total

Niidea([1,2,3,4])
    
```

```python
def MinMax(L):
    if len(L) == 1:
        min = L[0]
        max = L[0]
        return min, max
    
    min1, max1 = MinMax(L[:-1])
    a=L[-1]

    if a < min1:
        min1 = a
    if a > max1:
        max1 = a
    return min1, max1

MinMax([1,2,3,4,5,6,-9])
```

```python
def Picos(L):
    if len(L) < 3:
        return 0

    if L[1] > L[0] and L[1] > L[2]:
        return 1 + Picos(L[1:])
    else:
        return 0 + Picos(L[1:])

Picos([1,0,7,0,2,-23])
```

```python
def Contar(S, S1, a):
    contador = 0

    for i in range(len(S)):
        if S.top() == a:
            contador += 1
        S1.push(S.pop())
    
```
