```python
def find_min(T:LinkedBinaryTree):
    ''' Retorna el contenedor donde se encuentra el mínimo '''
    curr = T.root()
    while T.left(curr) is not None:
        curr = T.left(curr)
    return curr
```

```python
minimo = find_min(L)
print(minimo.element())
```

```python
def es_ordenado(T:LinkedBinaryTree):
    I = T.inorder()
    a = next(I)
    for i in range(len(T)-2):
        b = next(I)
        if b.element() < a.element():
            return False
        a = next(I)
        a,b= b,a
    return True
```
