```python
class cola_de_prioridad:
    def __init__(self):
        self.data =[]

    def __len__(self):
        return len(self.data)
    
    def is_empty(self):
        return len(self.data) == 0

    def add(self, k, v):
        self.data.append((k,v))

    def _indice_min(self):
        i = 0
        for j in range(1, len(self)):
            if self.data[j][0] < self.data[i][0]:
                i = j
            return i
    def min(self):
        return self.data[self._indice_min()]

    def remove_min(self):
        i = self._indice_min()
        return self.data.pop(i)
```

```python
CP = cola_de_prioridad()
for i in [2, -3, 5, 7]:
    CP.add(i, i**2)
```
